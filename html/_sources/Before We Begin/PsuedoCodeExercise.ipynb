In this exercise we will look at a two-dimensional random walk. A random walk is a mathematical model that describes a path consisting of a succession of random steps. In a 2D random walk, a particle starts at the origin and at each time step moves one unit in a randomly chosen direction (up, down, left, or right).

Random walks have applications in diffusion modeling, for example in physics (Brownian motion), biology (animal foraging patterns), and financial modeling (stock price movements). They help us understand how particles spread out over time and model various stochastic processes in nature.

Below is pseudocode for a two-dimensional random walk. Using the pseudocode, write a Python function to generate random walk data. 

**ALGORITHM:** Two-Dimensional Random Walk  
**INPUT:** n_steps (number of steps to take)  
**OUTPUT:** array of positions with shape (n_steps, 2)

```
// Initialize starting position and storage array
current_position = [0, 0]
positions = array of size (n_steps, 2)
positions[0] = current_position

// Take n_steps-1 additional steps
FOR i = 1 to n_steps-1
    // Generate random step in x and y directions
    step_x = random choice from [-1, 1]
    step_y = random choice from [-1, 1]
    
    // Update current position
    current_position[0] = current_position[0] + step_x
    current_position[1] = current_position[1] + step_y
    
    // Store the new position
    positions[i] = current_position
END FOR

RETURN positions
```

**Part 1: Single Random Walk Function**

```python
import numpy as np

def random_walk_2d(n_steps):
    """
    Generate a 2D random walk.
    
    Parameters:
    n_steps (int): Number of steps in the walk
    
    Returns:
    numpy.ndarray: Array of shape (n_steps, 2) containing positions
    """
    # Initialize starting position and storage array
    current_position = np.array([0, 0])
    positions = np.zeros((n_steps, 2))
    positions[0] = current_position
    
    # Take n_steps-1 additional steps
    for i in range(1, n_steps):
        # Generate random step in x and y directions
        step_x = np.random.choice([-1, 1])
        step_y = np.random.choice([-1, 1])
        
        # Update current position
        current_position[0] += step_x
        current_position[1] += step_y
        
        # Store the new position
        positions[i] = current_position.copy()
    
    return positions

# Test the function
np.random.seed(42)
walk = random_walk_2d(10)
print("Random walk positions:")
print(walk)
print(f"\nShape: {walk.shape}")
```

**Part 2: Multiple Random Walks Function**

```python
def multiple_random_walks(n_walks, n_steps):
    """
    Generate multiple 2D random walks.
    
    Parameters:
    n_walks (int): Number of different walks to generate
    n_steps (int): Number of steps in each walk
    
    Returns:
    numpy.ndarray: Array of shape (n_walks, n_steps, 2)
    """
    walks = np.zeros((n_walks, n_steps, 2))
    
    for walk_idx in range(n_walks):
        walks[walk_idx] = random_walk_2d(n_steps)
    
    return walks

# Generate multiple walks
np.random.seed(42)
multiple_walks = multiple_random_walks(5, 100)
print(f"Multiple walks shape: {multiple_walks.shape}")
print("Final positions of 5 walks:")
print(multiple_walks[:, -1, :])  # Last position of each walk
```

**Part 3: Analysis - Average Distance from Origin**

```python
def analyze_diffusion(walks):
    """
    Analyze the diffusion process by calculating average distance from origin.
    
    Parameters:
    walks (numpy.ndarray): Array of shape (n_walks, n_steps, 2)
    
    Returns:
    numpy.ndarray: Average distance from origin at each step
    """
    # Calculate distance from origin for each position
    distances = np.sqrt(walks[:, :, 0]**2 + walks[:, :, 1]**2)
    
    # Calculate average distance across all walks
    avg_distance = np.mean(distances, axis=0)
    
    return avg_distance, distances

# Analyze the diffusion
np.random.seed(123)
many_walks = multiple_random_walks(1000, 200)
avg_distances, all_distances = analyze_diffusion(many_walks)

print("Average distance from origin (first 10 steps):")
print(avg_distances[:10])
print(f"\nFinal average distance: {avg_distances[-1]:.2f}")
print(f"Standard deviation of final distances: {np.std(all_distances[:, -1]):.2f}")

# The average distance should grow approximately as sqrt(n_steps)
expected_final_distance = np.sqrt(200)
print(f"Theoretical expectation (sqrt(n_steps)): {expected_final_distance:.2f}")
```

**Bonus: Vectorized Implementation**

```python
def random_walk_2d_vectorized(n_steps):
    """
    Vectorized implementation for better performance.
    """
    # Generate all random steps at once
    steps = np.random.choice([-1, 1], size=(n_steps-1, 2))
    
    # Calculate cumulative sum to get positions
    positions = np.zeros((n_steps, 2))
    positions[1:] = np.cumsum(steps, axis=0)
    
    return positions

# Compare performance (for large n_steps)
import time

n_steps = 10000

# Original implementation
start = time.time()
walk1 = random_walk_2d(n_steps)
time1 = time.time() - start

# Vectorized implementation  
start = time.time()
walk2 = random_walk_2d_vectorized(n_steps)
time2 = time.time() - start

print(f"Original: {time1:.4f} seconds")
print(f"Vectorized: {time2:.4f} seconds")
print(f"Speedup: {time1/time2:.1f}x")
```

**Key Insights:**
- Random walks demonstrate diffusion: particles spread out over time
- The average distance from origin grows as √(number of steps)
- Vectorized NumPy operations are much faster for large datasets
- Multiple walks show the statistical nature of the diffusion process
