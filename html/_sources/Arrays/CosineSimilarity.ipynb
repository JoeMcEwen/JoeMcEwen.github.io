# Cosine Similarity: application of dot product 

**Cosine similarity** is a metric used to measure how similar two vectors are, regardless of their magnitude. It computes the cosine of the angle between two vectors in an inner product space. If the vectors are identical, the cosine similarity will be 1. If they are orthogonal (i.e., they have no direction in common), the cosine similarity will be 0. If they are opposite, the cosine similarity will be -1.

Cosine similarity is often used in **text mining**, **natural language processing**, and other fields where vector representations of items (e.g., words, documents) are common.

## Mathematical Definition

Given two vectors **A** and **B**, the cosine similarity is defined as:

$$
\text{Cosine Similarity}(\mathbf{A}, \mathbf{B}) = \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{A}\| \|\mathbf{B}\|}
$$

Where:
- $ \mathbf{A} \cdot \mathbf{B} $ is the **dot product** of the two vectors.
- $ \|\mathbf{A}\| $ and $ \|\mathbf{B}\| $ are the **magnitudes** (norms) of the vectors **A** and **B**.

In expanded form, the formula becomes:

$$
\text{Cosine Similarity}(\mathbf{A}, \mathbf{B}) = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \cdot \sqrt{\sum_{i=1}^{n} B_i^2}}
$$

Where:
- $ A_i $ and $ B_i $ are the components of vectors **A** and **B**, respectively.
- $ n $ is the number of dimensions (the number of elements in the vectors).

## Steps for Calculating Cosine Similarity

1. **Dot Product**: Compute the dot product of the two vectors.
2. **Norms**: Compute the norms (magnitudes) of both vectors.
3. **Divide**: Divide the dot product by the product of the magnitudes.

## Example Calculation

Suppose we have two vectors:

$$
\mathbf{A} = [1, 2, 3]
\quad \text{and} \quad \mathbf{B} = [4, 5, 6]
$$

1. **Dot product**:

$$
\mathbf{A} \cdot \mathbf{B} = (1)(4) + (2)(5) + (3)(6) = 4 + 10 + 18 = 32
$$

2. **Norms**:

$$
\|\mathbf{A}\| = \sqrt{(1)^2 + (2)^2 + (3)^2} = \sqrt{1 + 4 + 9} = \sqrt{14}
$$

$$
\|\mathbf{B}\| = \sqrt{(4)^2 + (5)^2 + (6)^2} = \sqrt{16 + 25 + 36} = \sqrt{77}
$$

3. **Cosine similarity**:

$$
\text{Cosine Similarity} = \frac{32}{\sqrt{14} \cdot \sqrt{77}} \approx 0.9746
$$

So, the cosine similarity between vectors **A** and **B** is approximately **0.9746**, which indicates a very high similarity between the two vectors.

## Applications of Cosine Similarity

- **Text Classification**: In natural language processing, cosine similarity is used to compare text documents based on the vector representations of the words or sentences.
- **Recommendation Systems**: In systems like movie recommendations or music playlists, cosine similarity can be used to measure how similar two users or items are based on their preferences or features.
- **Clustering**: In clustering algorithms, cosine similarity is used to measure the closeness of data points in high-dimensional spaces.

## Python Code Example

Here's how you can calculate cosine similarity in Python using **NumPy**:

```python
import numpy as np

# Example vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

# Cosine similarity calculation
dot_product = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)

cosine_similarity = dot_product / (norm_A * norm_B)
print("Cosine Similarity:", cosine_similarity)

**Example in Machine Learning**

We have two simple text documents, and we want to compare them using cosine similarity. We will convert the text into numerical vector representations using **[TF-IDF](https://en.wikipedia.org/wiki/Tf%E2%80%93idf)** (Term Frequency - Inverse Document Frequency) and then calculate the cosine similarity between the two vectors.

Here is the Python code. 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample text documents
doc1 = "I love machine learning and natural language processing."
doc2 = "Machine learning and language processing are fascinating."

# Create a TfidfVectorizer
vectorizer = TfidfVectorizer()

# Convert the text documents into vectors
tfidf_matrix = vectorizer.fit_transform([doc1, doc2])

# Compute the cosine similarity between the two documents
cos_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

# Print the cosine similarity
print(f"Cosine Similarity: {cos_sim[0][0]}")
```

Here are some more examples. 

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample text documents
doc1 = "I love machine learning and natural language processing."
doc2 = "Machine learning and language processing are fascinating."
doc3 = "I enjoy all things machine learning and natural language processing"
doc4 = "I heart machine learning with natural language processing"

# Create a TfidfVectorizer
vectorizer = TfidfVectorizer()

# Convert the text documents into vectors
tfidf_matrix = vectorizer.fit_transform([doc1, doc2, doc3, doc4])

# Compute the cosine similarity between the two documents
cos_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

# Print the cosine similarity
print(f"Cosine Similarity: {cos_sim[0][0]}")

# Compute the cosine similarity between the two documents
cos_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[2:3])

# Print the cosine similarity
print(f"Cosine Similarity: {cos_sim[0][0]}")

# Compute the cosine similarity between the two documents
cos_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[3:4])

# Print the cosine similarity
print(f"Cosine Similarity: {cos_sim[0][0]}")

# Compute the cosine similarity between the two documents
cos_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[0:1])

# Print the cosine similarity
print(f"Cosine Similarity: {cos_sim[0][0]}")
````
