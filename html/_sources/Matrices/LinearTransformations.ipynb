# Linear Transformations 

## What is a Linear Transformation?
A linear transformation takes an input vector and produces an output vector by applying a matrix multiplication followed by adding a bias vector. In machine learning, we use weights (W) and biases (b) to transform data from one space to another.

## Mathematical Definition

For an input vector **x** in $\mathbb{R}^n$, a linear transformation produces an output vector **y** in $\mathbb{R}^m$ using the equation

\begin{equation}
\mathbf{y} = \mathbf{W}\mathbf{x} + \mathbf{b}
\end{equation}

where:
- **W** is an $m \times n$ weight matrix
- **b** is an $m$-dimensional bias vector
- **x** is an $n$-dimensional input vector
- **y** is an $m$-dimensional output vector

**Example: 2D to 1D Transformation**

Consider transforming a 2D input to a 1D output. If we have input $\mathbf{x} = [x_1,~ x_2]^T$ and want output $y$, we use

\begin{equation}
y = w_1 x_1 + w_2 x_2 + b
\end{equation}

```python
import numpy as np

# Define weights and bias
W = np.array([[2, -1]])  # 1x2 matrix
b = np.array([3])        # 1D bias

# Input vector (2D)
x = np.array([4, 2])

# Apply linear transformation
y = W @ x + b
print(f"Input: {x}")
print(f"Output: {y}")  # Output: [9]
