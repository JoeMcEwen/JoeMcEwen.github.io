# Guassian Elimination 

```python
import numpy as np
```

```python
def forward_sub(L,b):
    
    n=b.size
    x=np.zeros(n)
    
    for i in range(n):
        x[i] = (b[i] - np.dot(L[i,:i],x[i+1:]))/L[i,i]
                
    return x

def backward_sub(U,b):
    
    n=b.size
    x=np.zeros(n) 
    
    x[n-1]=b[n-1]/U[n-1, n-1]
    
    for i in reversed(range(n-1)):
        
        x[i] = (b[i] - np.dot(U[i, i+1:],x[i+1:]))/U[i,i]
                
    return x

def gauss_elim(A_in, b_in):
    
    A=np.copy(A_in)
    b=np.copy(b_in) 
    
    n=b.size
    
    for j in range(n-1):
        for i in range(j+1,n):
            
            coef=A[i,j]/A[j,j]
            
            A[i,j:] -= coef*A[j,j:]
            b[i]-=coef*b[j] 
     
    print(A)
    print("b",b)
    return backward_sub(A,b)


A=np.random.uniform(0,10,size=(3,3))
x=np.random.uniform(0,10,size=3)
b=np.dot(A,x) 

print(b) 
print("x true", x)
print("x solved", gauss_elim(A,b))
```
