A symmetric matrix $A$ is said to be \textbf{positive definite} if for any non-zero vector $x$, we have:
\begin{equation}
    x^T A x > 0
\end{equation}

An example of a $2 \times 2$ positive definite symmetric matrix is:
\begin{equation}
    A = \begin{bmatrix}
    2 & 1 \\
    1 & 2
    \end{bmatrix}
\end{equation}

We can verify that $A$ is symmetric since $A^T = A$, and it is positive definite because:
\begin{equation}
    x^T A x = 
    \begin{bmatrix}
    x_1 & x_2
    \end{bmatrix}
    \begin{bmatrix}
    2 & 1 \\
    1 & 2
    \end{bmatrix}
    \begin{bmatrix}
    x_1 \\
    x_2
    \end{bmatrix}
    = 2x_1^2 + 2x_2^2 + 2x_1 x_2 > 0
\end{equation}
for all $x \neq 0$.
