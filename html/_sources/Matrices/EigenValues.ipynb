# EigenValues and Principle Component Analysis 

## The Eigen Value Equation
Eigen value equations are equations of the form 
\begin{equation} 
A \mathbf{v} = \lambda \mathbf{v} , 
\end{equation} 
where $A$ is a square matrix, $\mathbf{v}$ is a non-zero vector, called the eigen vector, and $\lambda$ is a scalar called the eigen value. 

The eigenvalue equation can be stated that $A$ acting on $\mathbf{v}$ returns a scaled version of $\mathbf{v}$, where the scaling factor is $\lambda$.
