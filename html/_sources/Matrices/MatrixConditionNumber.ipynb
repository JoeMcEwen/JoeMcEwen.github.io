A condition number measures how sensitive a function's output is to changes or errors in its input. In these notes, we focus on linear systems of equations, which are problems of the form:

\begin{equation}
A \mathbf{x} = \mathbf{b}.
\end{equation}

The question we will address is how the solution to equations of the form $A\mathbf{x} = \mathbf{b}$ changes if we introduce small perturbations to the elements of $A$ or $\mathbf{b}$. The **matrix condition number** quantifies the impact these small changes have on the solution.

Now, consider changing the elements of matrix $A$ slightly—how much does this affect the solution, $\mathbf{x}$? In a word-schematic equation, we can express this relationship as:

\begin{equation}
\text{changes in } \mathbf{x} \leq \text{some number} \times \text{small changes in } A.
\end{equation}

The matrix condition number is this "some number." If the number is small, then small changes in $A$ map to small changes in $\mathbf{x}$. However, if the number is large, small changes in $A$ may have a significant impact on the solution $\mathbf{x}$.

Likewise, the condition number specifies how small changes in $\mathbf{b}$ propagte to changes in $\mathbf{x}$.

## Matrix Norms
We will utilize the properties of [matrix norms](https://en.wikipedia.org/wiki/Matrix_norm). Below are the key properties of matrix norms:

- $|| A || > 0$
- $|| A || = 0$ if and only if all elements are zero, i.e., $A_{ij} = 0$ for all $i$, $j$
- $|| k A || = |k| || A ||$, where $k$ is a scalar
- $|| A + B || \leq || A || + || B ||$, known as the triangle inequality
- $|| A B || \leq || A || || B ||$

If you're wondering what a norm is, a common one is the Euclidean norm (also known as the $l^2$ norm or "el-2"), which is the square root of the sum of all elements squared:

\begin{equation}
|| A ||_2 = \sqrt{ \sum_i \sum_j A_{ij}^2}.
\end{equation}

You can check this in Python with the following code. 
```python
A = np.random.rand(3,3)
print(np.linalg.norm(A))
norm=0
for i in range(3):
    for j in range(3): 
        norm += A[i,j]**2
print(np.sqrt(norm))
```

## Condition Number 
We will now derive the condition number. Keep in mind that there are multiple ways to derive the condition number, and the approach we are using is just one of them.

We perturb the matrix $A$ by adding a small change, $\Delta A$, so that $A \to A + \Delta A$. As a result, we expect the output $\mathbf{x}$ to change under this perturbation. We should think of this as how the error introduced by adding $\Delta A$ to $A$ maps to an error in the solution, i.e., how $\mathbf{x}$ changes to $\mathbf{x} + \Delta \mathbf{x}$.

Adding perturbations to the equation $A \mathbf{x} = \mathbf{b}$, we have:

\begin{equation}
\begin{split} 
(A + \Delta A) ( \mathbf{x} + \Delta \mathbf{x}) & = \mathbf{b} \\
A \mathbf{x}  + \Delta A\mathbf{x} + A  \Delta \mathbf{x} + \Delta A \Delta \mathbf{x} & = \mathbf{b}.
\end{split} 
\end{equation} 
The first term on the left hand side and the right hand side are equal, since $A \mathbf{x} = \mathbf{b}$. Continuing we have
\begin{equation} 
\begin{split} 
 A  \Delta \mathbf{x} &  = - \Delta A\mathbf{x} -  \Delta A \Delta \mathbf{x} \\
A  \Delta \mathbf{x} &  = - \Delta A ( \mathbf{x} +  \Delta \mathbf{x}) \\
\Delta \mathbf{x} & = -A^{-1} \Delta A ( \mathbf{x} +  \Delta \mathbf{x}). 
\end{split} 
\end{equation} 
We now take the norm of both sides. 
\begin{equation} 
\begin{split} 
& ||\Delta \mathbf{x} || = ||A^{-1} \Delta A ( \mathbf{x} +  \Delta \mathbf{x}) || \\
& ||\Delta \mathbf{x} || \leq ||A^{-1} || ~||\Delta A ||~||( \mathbf{x} +  \Delta \mathbf{x}) || \\
& ||\Delta \mathbf{x} || \leq ||A^{-1} || ~||\Delta A ||~|| \mathbf{x}||
\end{split} 
\end{equation} 
Note, in the second line the norm property, $|| A +  B|| \leq || A|| + ||B||$ was used. Continuing we have
\begin{equation} 
\begin{split} 
& \frac{ || \Delta \mathbf{x} ||}{|| \mathbf{x} ||} \leq ||A^{-1} || ~||\Delta A || \frac{ || A ||}{||A||}  \\
& \frac{ || \Delta \mathbf{x} ||}{|| \mathbf{x} ||} \leq ||A^{-1} || ~|| A|| \frac{ ||\Delta A ||}{||A||}\\
& \frac{ || \Delta \mathbf{x} ||}{|| \mathbf{x} ||} \leq \kappa(A) \frac{ ||\Delta A ||}{||A||}, 
\end{split} 
\end{equation} 
where $\kappa(A) = ||A^{-1} || ~|| A||$ is the condition number. 
If the condition number is large, then even small perturbations in $A$ can lead to a large error in $\frac{ || \Delta \mathbf{x} ||}{|| \mathbf{x} ||}$. 

Likewise, the condition number sets a bound on how small changes in $\mathbf{b}$ propagate to changes in $\mathbf{x}$: \begin{equation} \frac{||\Delta \mathbf{x}||}{||\mathbf{x}||} \leq \kappa(A) \frac{||\Delta \mathbf{b}||}{||\mathbf{b}||}. \end{equation} This inequality shows that the relative change in the solution $\mathbf{x}$ is at most $\kappa(A)$ times the relative change in $\mathbf{b}$.



## What is a Large Condition Number
- condition numbers on the order of 1 are excellent 
- condition numbers on the order of $10$ to $10^2$ are considered "well-conditioned"
- condition numbers larger than $10^2$ are "ill-conditioned"

## How to get the Condition Number
Numpy provides a function for the condition number, np.linalg.cond().

### Hilbert Matrix

A [Hilbert matrix](https://en.wikipedia.org/wiki/Hilbert_matrix) $ H $ has elements defined by

\begin{equation} 
H_{ij} = \frac{1}{i + j - 1}.
\end{equation} 

A $3 \times 3$ Hilbert matrix is
\begin{equation}
H = 
\begin{bmatrix}
1 & \frac{1}{2} & \frac{1}{3} \\
\frac{1}{2} & \frac{1}{3} & \frac{1}{4} \\
\frac{1}{3} & \frac{1}{4} & \frac{1}{5}. 
\end{bmatrix}
\end{equation} 

Hilbert matrices are known to have large condition numbers. Below is an example of a $5 \times 5$ Hilbert matrix. You can use np.linalg.cond() to compute the condition number of the matrix.

### Exercise
Create a $5 \times 5$ Hilbert matrix and solve $H\mathbf{x} = \mathbf{b}$ for $\mathbf{b} = [1~ 1~ 1~ 1~ 1]^T$. 

````python
from scipy.linalg import hilbert

# Create a 5x5 Hilbert matrix
H = hilbert(5)

# Create a right-hand side vector (e.g., all ones)
b = np.ones(5)

# Solve Hx = b
x = np.linalg.solve(H, b)

# Calculate condition number
cond_number = np.linalg.cond(H)

print("Hilbert matrix (5x5):\n", H)
print("\nThis is vector b:", b) 
print("\nCondition number:", cond_number)
print("\nSolution x:", x)
print("\nCheck soltuion recovers b:", H@x) 
````


Let's change the vector $\mathbf{b}$ slightly by changing the first element to 0.1. Notice how much the solution $\mathbf{x}$ changes. 

````python
# Solve Hx = b
bp=np.copy(b)
bp[0] = .1
xp = np.linalg.solve(H, bp)

# Calculate condition number
cond_number = np.linalg.cond(H)

print("Hilbert matrix (5x5):\n", H)
print("\nThis is vector b:", bp) 
print("\nCondition number:", cond_number)
print("\nThe perturbed solution:", xp)
print("\nThe unperturbed solution:", x)
print("\nCheck soltuion recovers perturbed b:", H@xp) 
````

## Matrix 2 Norm, SVD, and Condition Number

### Matrix 2-Norm

The **2-norm** (also called the **spectral norm** or **operator norm**) of a matrix measures how much a matrix can stretch a vector in Euclidean space. It is defined as the maximum singular value of the matrix, and for square matrices, it corresponds to the largest eigenvalue in absolute value. A high 2-norm means the matrix can stretch vectors significantly, whereas a low 2-norm means the matrix causes little distortion.

For a matrix $A$, the **2-norm** is given by:

\begin{equation}
||A||_2 = \max_{\mathbf{x} \neq 0} \frac{||A\mathbf{x}||_2}{||\mathbf{x}||_2}
\end{equation}

where:
- $\mathbf{x}$ is any nonzero vector,
- $||A\mathbf{x}||_2$ is the 2-norm of $A\mathbf{x}$,
- $||\mathbf{x}||_2$ is the 2-norm of $\mathbf{x}$.

Alternatively, for square matrices, the 2-norm can be computed as:

\begin{equation}
||A||_2 = \sigma_{\max}(A)
\end{equation}

where $\sigma_{\max}(A)$ is the largest singular value of $A$, which can be computed via singular value decomposition (SVD).

---

### Properties of the 2-Norm

1. **Non-negativity**: $||A||_2 \geq 0$, and $||A||_2 = 0$ if and only if $A$ is the zero matrix.
2. **Scaling**: For any scalar $k$, $||kA||_2 = |k| \, ||A||_2$.
3. **Triangle inequality**: $||A + B||_2 \leq ||A||_2 + ||B||_2$.
4. **Sub-multiplicative property**: $||AB||_2 \leq ||A||_2 ||B||_2$.

---

### Singular Values

The **SVD** (singular value decomposition) of a matrix $A$ is:

\begin{equation}
A = U \Sigma V^T,
\end{equation}

where:
- $U \in \mathbb{R}^{m \times m}$ is orthogonal ($U^T U = \mathbf{I}$),
- $\Sigma \in \mathbb{R}^{m \times n}$ is diagonal with non-negative entries,
- $V \in \mathbb{R}^{n \times n}$ is orthogonal ($V^T V = \mathbf{I}$).

The diagonal entries $\sigma_1, \sigma_2, \dots, \sigma_r$ of $\Sigma$ are called the **singular values** of $A$. Singular values describe how much $A$ stretches vectors along different directions.

- $\sigma_1 \geq \sigma_2 \geq \dots \geq \sigma_r \geq 0$,
- $\sigma_1$ = maximum stretching,
- $\sigma_r$ = minimum stretching.

---

### Matrix Condition Number from SVD

The **condition number** of $A$ (with respect to the 2-norm) is defined as:

\begin{equation}
\kappa(A) = \frac{\sigma_{\max}(A)}{\sigma_{\min}(A)}
\end{equation}

where:
- $\sigma_{\max}(A)$ = largest singular value,
- $\sigma_{\min}(A)$ = smallest nonzero singular value.

---

### Example 
Below the condition number and matrix norm for a $6 \times 6$ Hilbert matrix is computed via SVD and shown to match np.linalg.cond() and np.linalg.norm(). 

````python
# 6 by 6 Hilbert matrix 
H = hilbert(6)
print("6 by 6 Hilbert matrix:\n",H)
````

````python
print("Condition number from linalg.cond:",np.linalg.cond(H))
````

````python
print("2-norm from linalg.norm:",np.linalg.norm(H,2))
````

````python
# SVD factorization
U,S,VT=np.linalg.svd(H)
````

````python
print("Max singular value, compare to 2-norm")
np.max(S)
````

````python
# remove any zero elements in S
idx=np.nonzero(S)
S=S[idx]
print("Ratio of maximum and minimum (smallest non-zero) singular values, compare to condition number.") 
np.max(S)/np.min(S)
````
