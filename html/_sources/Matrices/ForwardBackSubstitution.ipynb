# Forward and Backward Substitution 

*If a matrix is triangular, then solving a systems of equatons is, well, "not so bad."*

A lower triangular matrix has non-zero elements on the diagonal and below it, but all elements above the diagonal are zero. Here's an example of a $3\times3$ lower triangular matrix:

\begin{equation} 
L = \begin{bmatrix}
l_{11} & 0     & 0     \\
l_{21} & l_{22} & 0     \\
l_{31} & l_{32} & l_{33}
\end{bmatrix}.
\end{equation} 

An upper triangular matrix has non-zero elements on the diagonal and above it, but all elements below the diagonal are zero. Here's an example of a $3\times3$ upper triangular matrix:

\begin{equation} 
U = \begin{bmatrix}
u_{11} & u_{12} & u_{13} \\
0      & u_{22} & u_{23} \\
0      & 0      & u_{33}
\end{bmatrix}
\end{equation} 

## Forward Substitution 

Given the system $L \mathbf{x} = \mathbf{b}$, where $L$ is a lower triangular matrix, we aim to solve for the vector $\mathbf{x}$. The matrix $L$ is of the form:

\begin{equation}
L = \begin{bmatrix}
\ell_{11} & 0         & \cdots & 0 \\
\ell_{21} & \ell_{22} & \cdots & 0 \\
\vdots    & \vdots    & \ddots & \vdots \\
\ell_{n1} & \ell_{n2} & \cdots & \ell_{nn}
\end{bmatrix}
\end{equation}

The system of equations is:

\begin{equation}
L \mathbf{x} = \mathbf{b}
\end{equation}

This gives the following equations:

\begin{equation}
\ell_{11} x_1 = b_1
\end{equation}
\begin{equation}
\ell_{21} x_1 + \ell_{22} x_2 = b_2
\end{equation}
\begin{equation}
\ell_{31} x_1 + \ell_{32} x_2 + \ell_{33} x_3 = b_3
\end{equation}
\begin{equation}
\vdots
\end{equation}
\begin{equation}
\ell_{n1} x_1 + \ell_{n2} x_2 + \cdots + \ell_{nn} x_n = b_n
\end{equation}

We solve for $x$ starting from the first equation and moving downward.

**Step 1: Solve for $x_1$**

\begin{equation}
\ell_{11} x_1 = b_1
\end{equation}

\begin{equation}
x_1 = \frac{b_1}{\ell_{11}}
\end{equation}

**Step 2: Solve for $x_2$**

\begin{equation}
\ell_{21} x_1 + \ell_{22} x_2 = b_2
\end{equation}

Substitute $x_1$:

\begin{equation}
x_2 = \frac{b_2 - \ell_{21} x_1}{\ell_{22}}
\end{equation}

**Step 3: General Step for $x_i$**

For $i = 1, 2, \dots, n$, we solve:

\begin{equation}
\ell_{i1} x_1 + \ell_{i2} x_2 + \cdots + \ell_{ii} x_i = b_i
\end{equation}

Solving for $x_i$:

\begin{equation}
x_i = \frac{b_i - \sum_{j=1}^{i-1} \ell_{ij} x_j}{\ell_{ii}}
\end{equation}

**Final Formula**

The general formula for each $x_i$ using forward substitution is:

\begin{equation}
x_i = \frac{b_i - \sum_{j=1}^{i-1} \ell_{ij} x_j}{\ell_{ii}}, \quad i = 1, 2, \dots, n
\end{equation}

Here is a Python code for forward substitution. 

```python
def forward_sub(L, b):
    """
    Solves Ly = b for y using forward substitution,
    where L is a lower triangular matrix.
    """
    n = L.shape[0]
    y = np.zeros_like(b, dtype=float)

    for i in range(n):
        if L[i, i] == 0:
            raise ValueError("Matrix is singular.")
        y[i] = (b[i] - np.dot(L[i, :i], y[:i])) / L[i, i]

    return y

```

## Backward Substitution 
Given the system $U \mathbf{x} = \mathbf{b}$, where $U$ is an upper triangular matrix, we aim to solve for the vector $\mathbf{x}$. The matrix $U$ is of the form:

\begin{equation}
U = \begin{bmatrix}
u_{11} & u_{12} & \cdots & u_{1n} \\
0 & u_{22} & \cdots & u_{2n} \\
0 & 0 & \ddots & \vdots \\
0 & 0 & 0 & u_{nn}
\end{bmatrix}
\end{equation}

The system of equations is:

\begin{equation}
U \mathbf{x} = \mathbf{b} 
\end{equation}

This leads to the following system of equations:

\begin{equation}
u_{11} x_1 + u_{12} x_2 + \cdots + u_{1n} x_n = b_1
\end{equation}
\begin{equation}
u_{22} x_2 + \cdots + u_{2n} x_n = b_2
\end{equation}
\begin{equation}
\vdots
\end{equation}
\begin{equation}
u_{nn} x_n = b_n
\end{equation}

We solve for $x$ starting from the last equation and moving upwards.

**Step 1: Solve for $x_n$**

From the last equation:

\begin{equation}
u_{nn} x_n = b_n
\end{equation}

Thus:

\begin{equation}
x_n = \frac{b_n}{u_{nn}}
\end{equation}

**Step 2: Solve for $x_{n-1}$**

\begin{equation}
u_{n-1, n-1} x_{n-1} + u_{n-1, n} x_n = b_{n-1}
\end{equation}

Substitute $x_n$:

\begin{equation}
x_{n-1} = \frac{b_{n-1} - u_{n-1, n} x_n}{u_{n-1, n-1}}
\end{equation}

**Step 3: General Step for $x_i$**

For each $i = n-2, n-3, \dots, 1$, we solve:

\begin{equation}
u_{ii} x_i + u_{i, i+1} x_{i+1} + \cdots + u_{i, n} x_n = b_i
\end{equation}

Solving for $x_i$:

\begin{equation}
x_i = \frac{b_i - \sum_{j=i+1}^{n} u_{ij} x_j}{u_{ii}}
\end{equation}

**Final Formula**

The general formula for backward substitution is:

\begin{equation}
x_i = \frac{b_i - \sum_{j=i+1}^{n} u_{ij} x_j}{u_{ii}}, \quad i = n, n-1, \dots, 1
\end{equation}

Here is a Python code for backward substitution. 

```python
def back_sub(U, y):
    """
    Solves Ux = y for x using back substitution,
    where U is an upper triangular matrix.
    """
    n = U.shape[0]
    x = np.zeros_like(y, dtype=float)

    for i in reversed(range(n)):
        if U[i, i] == 0:
            raise ValueError("Matrix is singular.")
        x[i] = (y[i] - np.dot(U[i, i+1:], x[i+1:])) / U[i, i]

    return x
```
